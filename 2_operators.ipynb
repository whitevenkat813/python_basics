# **Python - operator**

* Arthmatic operators - +, -, *, /, **, //, %

* Assignment operators - =, +=, -=, *=, /=, **=, //=, %=

* Comparison operators - ==, !=, <, >, <=, >=

* Logical operators - and, or, not

* Identity operators - is, is not

* Membership operators - in, not in




```python
#  Arthmatic operators - +, -, *, /, **, //, %

print(10 + 2)
print(10 - 2)
print(10 * 2)
print(10 / 2)
print(10 ** 2)
print(10 // 2)
print(10 % 2)
```

```python
# Assignment operators - =, +=, -=, *=, /=, **=, //=, %=

a = 20
print(a)

a += 5
print(a)    

a -= 5
print(a)        

a *= 5
print(a)    

a /= 5
print(a)    

a **= 5
print(a)

a //= 5
print(a)    

a %= 5
print(a)

```

```python
# Comparison operators - ==, !=, <, >, <=, >=  # True or False

print(10 == 10) #True
print(10 != 10) #False
print(10 < 10) #False
print(10 > 10) #False
print(10 <= 10) #True
print(10 >= 10) #True
```

```python
# logical operators - and, or, not  # True or False


print("-----and-----")
print(True and True)  # True
print(True and False)  # False  
print(False and True)  # False
print(False and False)  # False
print("-----or-----")
print(True or False)  # True
print(False or True)  # True
print(False or False)  # False
print("-----not-----")
print(not True)  # False
print(not False)  # True
```

```python
print("-----and-----")
print(10 > 5 and 10 < 20)  # True
print(10 > 5 and 10 > 20)  # False  
print(10 < 5 and 10 < 20)  # False
print(10 < 5 and 10 > 20)  # False

print("-----or-----")
print(10 > 5 or 10 < 20)  # True
print(10 > 5 or 10 > 20)  # True
print(10 < 5 or 10 < 20)  # True
print(10 < 5 or 10 > 20)  # False   

print("-----not-----")
print(not(10 > 5))  # False
print(not(10 < 5))  # True

```
