# **python data types**

* integer => 1, 1234, -1234 
* float => 1.0, 1234.0, -1234. 
* complex => 1+2j, 1.0+2.0j 
* boolean => True, False 
* string => 'hello', "hello", 



```python
# * integer => 1, 1234, -1234

a = 10
a1 = -20
print(a, type(a))
print(a1, type(a1))

# float => 1.0, 1.23, -1.23
b = 10.5
b1 = -20.5
print(b, type(b))
print(b1, type(b1))

# complex => 1+2j, 1-2j, -1+2j, -1-2j

c = 1 + 2j
c1 = 1 - 2j
c2 = -1 + 2j
c3 = -1 - 2j
print(c, type(c))
print(c1, type(c1))
print(c2, type(c2))
print(c3, type(c3))


# string => "hello", 'hello', """hello""", '''hello'''


d = "hello"
d1 = 'hello'
d2 = "10"
d3 = "30" + "20"
print(d, type(d))
print(d1, type(d1))
print(d2, type(d2))
print(d3, type(d3))

# boolean => True, False

e = True
e1 = False
print(e, type(e))
print(e1, type(e1))
```

# **escape sequences**

* \ - for new line in editor
* \\ - print \
* \' - print '
* \" - print "
* \n - print a new line
* \t - print a tap


```python
print("hello\\world\\")
print("10\\100")
print("I\'am pyton developer")
print("Hello\"world\"")
print("Hello\nworld")
print("Hello\t world")
```
