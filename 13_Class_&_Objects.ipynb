# **Class & objects**

```python
class vehicle:
    x = 10 # class variable
    y = 30 # class variable

obj1 = vehicle()

print(obj1.x,obj1.y)

print("--------------------")

print(obj1.x)
print(obj1.y)

```

```python
class vehicle:
    x = 10 # class variable
    y = 30 # class variable

obj1 = vehicle()
obj2 = vehicle()


print(obj1.x)
print(obj1.y)

obj1.x = "venkat"
print(obj1.x,obj1.y)

print("--------------------")

print(obj2.x,obj2.y) # Accessing class variables through an object


```

```python
class vehicle:
    x = 10 # class variable
    y = 30 # class variable

    def myfunc(self):     # self parameter is used to access the class variables
        print("Hello my name is venkat")

        print(self)       # self is passed as an object to the function

obj = vehicle()
print(obj.x,obj.y)

obj.myfunc()

print(obj)
```

# **constructor - function**

* syntax:
         \__init\__\()


```python
class xyz:

    def __init__(self,a,b):     # a, b are function variables
        print("constructor function is called")
        self.a = a
        self.b = b


obj1 = xyz(100,200)
print(obj1.a,obj1.b)

obj2 = xyz("venkat","jeeva")
print(obj2.a,obj2.b)


```

```python
class studend:

    def __init__(self,Name,Age,Place):
        print("the studend bio deta :")

        self.Name = Name
        self.Age = Age
        self.Place = Place

venkat = studend("venkat",25,"nagapattinam")
print(venkat.Name)
print(venkat.Age)
print(venkat.Place)
    
print("---------------------")

Tamil = studend("kural",27,"nagapattinam")
print(Tamil.Name)
print(Tamil.Age)
print(Tamil.Place)
    

```
